```python
import pandas as pd
from nltk import word_tokenize, pos_tag
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer

# Function to lemmatize words
def lemmatize_words(text):
    if pd.isnull(text):
        return ''
    lemmatizer = WordNetLemmatizer()
    lemmatized_words = []
    for word, pos in pos_tag(word_tokenize(text)):
        wordnet_pos = pos[0].lower()
        wordnet_pos = 'n' if wordnet_pos == 'j' else wordnet_pos
        wordnet_pos = 'v' if wordnet_pos == 'r' else wordnet_pos
        if wordnet_pos not in ['a', 'r', 'n', 'v']:
            wordnet_pos = 'n'
        lemma = lemmatizer.lemmatize(word, pos=wordnet_pos)
        lemmatized_words.append(lemma)
    return ' '.join(lemmatized_words)

# Apply lemmatization to each column
df['keywords_description'] = df['keywords_description'].apply(lambda x: lemmatize_words(x.lower()))
df['keywords'] = df['keywords'].apply(lambda x: lemmatize_words(x.lower()))
df['term'] = df['term'].apply(lambda x: lemmatize_words(str(x).lower()))  

```

```python



import spacy

# Load the pre-trained spaCy model
nlp = spacy.load("en_core_web_sm")

# Define a set of stopwords
stopwords = ["video","body","black"]

# Function to extract nouns excluding stopwords
def extract_nouns(text):
    doc = nlp(text)
    keywords = [token.text for token in doc if token.pos_ == "NOUN" and token.text.lower() not in stopwords]
    return keywords
        
df['new_keywords'] = df['new_description'].apply(extract_nouns)



df['keywords'] = df['keywords'].apply(literal_eval)
df['keywords_description'] = df['keywords_description'].apply(literal_eval)

import nltk
from nltk.stem import WordNetLemmatizer

# Download WordNet if not already downloaded
nltk.download('wordnet')

# Initialize the WordNet lemmatizer
lemmatizer = WordNetLemmatizer()

# Function to lemmatize a list of words
def lemmatize_words(words):
    return [lemmatizer.lemmatize(word) for word in words]

# Assuming df['new_keywords'] contains lists of words, apply lemmatization
df['new_keywords'] = df['new_keywords'].apply(lemmatize_words)



import pandas as pd
from collections import Counter
from ast import literal_eval
import nltk
from nltk.stem import WordNetLemmatizer

def lemmatize_words(words):
    return [lemmatizer.lemmatize(word) for word in words]

# Function to find most frequent words and update missing ones
def process_keywords(row):
    if row["source"]=="pixabay":
        all_keywords = row['keywords'] + row['new_keywords']
    if row["source"]=="unsplash":
        all_keywords = row['keywords_description'] + row['new_keywords']
    else :
        all_keywords = row['keywords'] + row['keywords_description'] + row['new_keywords']

    normalized_keywords = lemmatize_words(all_keywords)
    word_counts = Counter(normalized_keywords)
    most_common_words = word_counts.most_common(5)  # Adjust as needed
    most_common_words_only = [word for word, _ in most_common_words]
    
    # Check and update missing words
    new_keywords_set = set(row['new_keywords'])
    current_most_common_set = set(most_common_words_only)
    missing_words = new_keywords_set - current_most_common_set
    
    if missing_words:
        most_common_words_only.extend(missing_words)
    
    return most_common_words_only

# Apply the function to each row
df['most_frequent_words'] = df.apply(process_keywords, axis=1)
```
