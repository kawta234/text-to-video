```python
import ollama
def get_model_response(prompt):
    response = ollama.generate(model='llama3',prompt=prompt,)
    return response['response']

def extract_main_topics(response):
    topics = [line.split('**')[1].strip() for line in response.split('\n') if line.startswith('1.') or line.startswith('2.') or line.startswith('3.')]
    return topics

prompt = "what are the 3 main topics in this text : A road in the middle of a snowy forest"
response = get_model_response(prompt)
print(response)

topics = extract_main_topics(response)
print(topics)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")


def extract_nouns(text):
    doc = nlp(text)
    keywords = [token.text for token in doc if token.pos_ == "NOUN" and token.text.lower()]
    return keywords

extract_nouns("A road in the middle of a snowy forest")
```

```python
import spacy
from keybert import KeyBERT

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Initialize KeyBERT model
kw_model = KeyBERT()

# Input sentence
sentence = "Join us in our vision to create a world where everyone can achieve optimal health and well-being"
# Extract keywords
keywords = kw_model.extract_keywords(sentence, keyphrase_ngram_range=(1, 1), stop_words='english', top_n=10)

# Filter only nouns
doc = nlp(sentence)
nouns = [token.text for token in doc if token.pos_ == "NOUN"]

# Get top 2 noun keywords
noun_keywords = [keyword[0] for keyword in keywords if keyword[0] in nouns][:2]

# Print the extracted noun keywords
for noun_keyword in noun_keywords:
    print(noun_keyword)
```
